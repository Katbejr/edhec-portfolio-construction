```python
data = pd.read_csv(
    "../data/Portfolios_Formed_on_ME_monthly_EW.csv",
    header=0,
    index_col=0,
    na_values=-99.99
)

data.index = pd.to_datetime(data.index, format="%Y%m").to_period("M")
```

```python
rets = data[["Lo 20", "Hi 20"]]
```

```python
rets = rets / 100
n_months = rets.shape[0]

annualized_rets = (1 + rets).prod() ** (12 / n_months) - 1
annualized_rets
```

```python
volatility = rets.std(ddof=0) * np.sqrt(12)
volatility
```
