```python
import sys
sys.path.append('..')  # pour importer portfolio_analytics depuis la racine
%load_ext autoreload
%autoreload 2
%matplotlib inline
import pandas as pd
import portfolio_analytics as erk
import numpy as np
```

```python
ind_return = erk.get_ind_returns()
ind_nfirms = erk.get_ind_nfirms()
ind_size = erk.get_ind_size()
```

```python
ind_mkcap = ind_nfirms * ind_size
ind_mkcap.shape
```

```python
total_mkcap = ind_mkcap.sum(axis='columns')
```

```python
total_mkcap.plot(figsize=(12,6))
```

```python
ind_capweight = ind_mkcap.divide(total_mkcap, axis="rows")
ind_capweight.index = pd.to_datetime(
        ind_capweight.index.astype(str), format="%Y-%m"
    ).to_period("M")
```

```python
ind_capweight.head()
```

```python
ind_capweight["1926"].sum(axis = 'columns')
```

```python
ind_capweight[['Fin','Steel']].plot(figsize=(12,6))
```

```python
total_market_return = (ind_capweight*ind_return).sum(axis= 'columns')
```

```python
total_market_return.plot()
```

```python
total_market_index = erk.drawdown(total_market_return).Wealth
total_market_index.plot(figsize=(10,5), title = "Total Market CapWeighted Index 1926-2018")
```

```python
total_market_index.loc['1980':].plot()
total_market_index.loc['1980':].rolling(window=36).mean().plot(figsize=(12,6))
```

```python
tmi_tr36rets = total_market_return.rolling(window=36).aggregate(erk.annualize_rets, periods_per_year=12)
tmi_tr36rets.plot(figsize=(12,6))
#total_market_return.plot(label="Return", legend=True)
```

```python
 ### Rolling Correlation - alomng with MultiIndez and .groupby
```

```python
ts_corr = ind_return.rolling(window=36).corr()
ts_corr.tail()
```

```python
ind_return.head()
```

```python
ts_corr.index.names = ['date','industry']
ts_corr.tail()
```

```python
ind_tr36corr = ts_corr.groupby(level='date').apply(lambda cormat: cormat.values.mean())
```

```python
ind_tr36corr.plot()
```

```python
tmi_tr36rets.plot(label="Tr36 Mo Rets", figsize=(12,6),legend =True)
ind_tr36corr.plot(label="Tr36 Mo Rets",legend =True, secondary_y=True)
```
