```python
def discount(t, r):
    ''''
    Compute the price of a pure discount bond that pays a dollar at time t, given interest rate t
    '''
    return (1+r)**(-t)
```

```python
def pv(l, r):
    '''
    computess the present value of sequence of liabilities 
    l is indexed by the time, and the values are the amounts of each liability
    returns the presnet value of the sequence
    '''
    dates = l.index
    discounts = discount(dates, r)
    return (discounts*l).sum()
```

```python
def funding_ratio(assets, liabilites, r):
    return assets/pv(liabilites, r)
```

```python
# Définir les liabilities comme une pandas Series indexée par le temps
import pandas as pd

liabilities = pd.Series(data=[1, 1.5, 2, 2.5], index=[3, 3.5, 4, 4.5])
def show_funding_ratio(assets, r):
    fr = funding_ratio(assets, liabilities, r)  # liabilities fixe, pas un widget
    print(f'{fr*100:.2f}')

controls = widgets.interactive(show_funding_ratio,
                               assets = widgets.IntSlider(min=1, max=10, step=1, value=5),
                               r      = (0, .20, 0.01)
                               )
display(controls)
```
