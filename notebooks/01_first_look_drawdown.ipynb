```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
me_m = pd.read_csv("../data/Portfolios_Formed_on_ME_monthly_EW.csv",
                   header=0, index_col=0, parse_dates=True, na_values=-99.99,date_format="%Y%m")
```

```python
rets = me_m[['Lo 10', 'Hi 10']]
rets.columns = ['Smallcaps', 'Highcaps']
rets = rets/100
rets.plot.line()
```

```python
rets.index = rets.index.to_period('M')
rets.head()
```

```python
rets.info()
rets.describe()
```

```python
wealth_index = 1000*(1+rets["Highcaps"]).cumprod()
wealth_index.plot()
```

```python
previous_peak = wealth_index.cummax()
previous_peak.plot()
```

```python
drawdown= (wealth_index-previous_peak)/previous_peak
drawdown.plot()
```
